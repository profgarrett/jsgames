## 1. Setup & Sample Data

Run the cell below to import libraries and create some example tables.

```python
import pandas as pd
import numpy as np

# Customers table
customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4, 5],
    'name': ['Alice', 'Bob', 'Carlos', 'Diana', 'Eve'],
    'city': ['Morgantown', 'Charleston', 'Morgantown', 'Huntington', 'Charleston'],
    'loyalty_level': ['Gold', 'Silver', 'Gold', 'Bronze', 'Silver']
})

# Orders table
orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104, 105, 106, 107],
    'customer_id': [1, 1, 2, 3, 3, 4, 5],
    'product': ['Bike', 'Helmet', 'Helmet', 'Bike', 'Gloves', 'Gloves', 'Bike'],
    'quantity': [1, 2, 1, 1, 3, 2, 1],
    'price_per_unit': [500, 50, 50, 500, 20, 20, 500]
})

orders['total_price'] = orders['quantity'] * orders['price_per_unit']

print('Customers:')
display(customers)
print('Orders:')
display(orders)
```

## 2. Filtering Rows

Filtering lets you keep only the rows that meet a condition.

### Demo: Filter orders with total price > 100


```python
high_value_orders = orders[orders['total_price'] > 100]
high_value_orders
```

## 4. Copy a table

We can create a new copy of a table to make changes.

```python
orders_copy = orders.copy()
# We can make changes to the new copy without modifying the original.
orders_copy = orders_copy.drop(columns = ['order_id', 'customer_id', 'quantity'])

print(f'New table has {len(orders_copy.columns)} columns.')
print(f'Original table has {len(orders.columns)} columns.')
```

## 5. Mutating / Creating New Columns

You can create or modify columns using assignment or `.assign()`.

### Demo: Add discount and final price


```python
orders_with_discount = orders.assign(
    discount_rate = np.where(orders['product'] == 'Bike', 0.10, 0.0)
)
orders_with_discount['discount_amount'] = orders_with_discount['total_price'] * orders_with_discount['discount_rate']
orders_with_discount['final_price'] = orders_with_discount['total_price'] - orders_with_discount['discount_amount']
orders_with_discount
```

## 6. Grouping Rows

`groupby` lets you split data into groups based on one or more columns.

### Demo: Group orders by product


```python
orders_by_product = orders.groupby('product')
for name, group in orders_by_product:
    print(f'Product: {name}')
    display(group)
    print('\n')
```

## 7. Aggregating

Aggregations summarize data, such as sums, counts, and averages.
Use `.agg()` or shortcut methods like `.sum()` or `.mean()`.

### Demo: Total quantity and revenue per product


```python
product_summary = (
    orders
    .groupby('product', as_index=False)
    .agg(
        total_quantity=('quantity', 'sum'),
        total_revenue=('total_price', 'sum'),
        avg_price=('price_per_unit', 'mean')
    )
)
product_summary
```
