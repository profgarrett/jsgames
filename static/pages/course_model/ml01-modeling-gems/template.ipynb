# ML01: Mining for Gems


## Task

You have a mining company looking for valuable gems. There are two different locations, one better than the other.

Each time you dig, you may find either a gem or a worthless rock. It costs you $1 to dig in a location. You can dig in either of two locations, which we will call Jar A and Jar B.

How many times do you dig in each location before deciding on which jar is the best?

Your goal is to pick the smallest number of samples that results in the right decision 99% of the time.

For reference, see the [Arkansas Craters of Diamonds State Park](https://www.arkansas.com/murfreesboro/accommodations/crater-diamonds-state-park)

```python
import numpy as np

# Create two jars with binary outcomes (1 = success, 0 = failure)
# You can add the data points to each jar below by typing numbers between the brackets, separated by commas.
# Eg: jar_a = [1, 0, 1, 1, 0]
jar_a = [1, 0, 1, 1, 0]
jar_b = [0, 0, 1, 1, 0]

# Make sure that jar_a is larger than jar_b
if len(jar_a) < len(jar_b):
    raise ValueError("jar_a must be larger than jar_b")

# Which is the best jar?
print('Mean of jar_a:', np.mean(jar_a) )
print('Mean of jar_b:', np.mean(jar_b) ) 

# Calcluate differnt between means

```

```python
# Use a two-sample t-test test to determine if the difference in means is significant
from scipy.stats import ttest_ind

# Testing to see if the means of the two jars are significantly different
result = ttest_ind(jar_a, jar_b) # compare against jar b

print('p-value:', result.pvalue)
print('confidence interval:', result.confidence_interval(0.95))

```

```python
# But how consistent are the results? We will randomly generate data 

# Set random seed for reproducibility
# np.random.seed(42)

# how many samples should we simulate in each jar?
sample_size = 10

# Use a binomial distribution to simulate the jars (1 or 0)
# n is number of trials (1 for each sample), p is probability of success (mean of jar), size is number of samples
jar_a_simulated = np.random.binomial(n=1, p=np.mean(jar_a), size=sample_size) 
jar_b_simulated = np.random.binomial(n=1, p=np.mean(jar_b), size=sample_size)

# Contents of the simulated jars
print('New jar_a (simulating ', np.mean(jar_a), ', actual ', np.mean(jar_a_simulated), '):', jar_a_simulated) 
print('New jar_b (simulating ', np.mean(jar_b), ', actual ', np.mean(jar_b_simulated), '):', jar_b_simulated)

result = ttest_ind(jar_a_simulated, jar_b_simulated)

# Print the results of the t-test, getting the t-statistic, confidence interval, and p-value
# The ci shows the difference in means with a 95% confidence level
# The p-value indicates the probability of observing the data if the null hypothesis is true (no difference in means)
print('p-value:', result.pvalue)
print('confidence interval:', result.confidence_interval(0.95))
```

```python
# Let's try multiple experiments to see how often we get the correct result
num_experiments = 100 
sample_size = 2 # how many should we pull each time?
correct_results = [] # empty array

# Simulate new jars for each experiment
for experiment_number in range(num_experiments):
    np.random.seed(experiment_number)  # Ensure reproducibility for each experiment

    jar_a_simulated = np.random.binomial(n=1, p=np.mean(jar_a), size=sample_size) 
    jar_b_simulated = np.random.binomial(n=1, p=np.mean(jar_b), size=sample_size) 

    # Did we get the right result, where our simulated results matches our original jars?
    if(sum(jar_a_simulated) > sum(jar_b_simulated) and sum(jar_a) > sum(jar_b)): 
        correct_results.append(1)
    else:
        correct_results.append(0)

# Analyze the results of the experiments
print(correct_results)

 # show the proportion of times we had the correct results (jar b > jar a)
print('Mean correct results over experiments:', np.mean(correct_results))
```
