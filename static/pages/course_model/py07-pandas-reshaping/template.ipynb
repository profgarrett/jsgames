## Changing Values

We often need to fix fields.

- Adding a field with `df['new_field'] = value` where value can be a constant or derived from other columns.
- Removing a field with `df.drop(columns=['field_to_remove'])`.
- Renaming a field with `df.rename(columns={'old_name': 'new_name'})`.

```python
# Sample changing values
import pandas as pd

# Create sample data
df_employees = pd.DataFrame({
    'employee_id': [101, 102, 103, 104],
    'name': ['Emma', 'James', 'Sofia', 'Michael'],
    'department': ['Engineering', 'Marketing', 'Sales', 'HR'],
    'salary': [75000, 65000, 58000, 72000],
    'hobby': ['Reading', 'Gaming', 'Travel', 'Sports']
})

# Add a new field 'high_earner' based on salary

print("After adding 'high_earner' field:")
print(df_employees)

# Remove the 'hobby' field

print("\nAfter removing 'hobby' field:")
print(df_employees)

# Rename 'salary' to 'annual_salary'

print("\nAfter renaming 'salary' to 'annual_salary':")
print(df_employees)

# Create a 0/1 binary field for engineering department.

print("\nAfter creating 'is_engineering' field:")
print(df_employees)
```

## Shaping data

Our data can be in two main formats: tall or wide. Tall is better for charting, but wide is easier to read and better for most statistical analyses. We can convert between the two formats.

*Wide → Tall (long)*  with `pd.melt()`

Arguments:

- `id_vars = ['id', 'name']`: a list of columns to keep. Any column not present will be melted.
- `value_vars = ['math', 'english']`: Optional! Columns to melt. If not given, will melt all columns not in id_vars.
- `var_name = 'subject'`: name for the new *variable* column
- `value_name = 'score'`: name for the new *value* column

*Tall → Wide*  with `pivot()`

Arguments:

- `index = ['id', 'name']`: columns to keep as index for each row
- `columns = 'subject'`: column(s) to use as the new field names
- `values = 'score'`: column(s) to fill use as values in the new fields

## Join data

Arguments:

- `on`: column(s) to join on. If not specified, pandas will use columns with the same name in both tables.
- `left_on`: column(s) from the left DataFrame to join on.
- `right_on`: column(s) from the right DataFrame to join on.
- `suffixes`: tuple of string suffixes to apply to overlapping column names in the left and right DataFrames.

## Grouping Data

We can group data by one or more columns and then perform aggregate functions on the grouped data. This is done using the `groupby()` and `agg()` functions.

*Group by* with `groupby()`

Arguments:
- `by`: column(s) to group by.
- `as_index`: if True, the grouped columns will be set as the index of the resulting DataFrame. Default is True.
- `sort`: if True, the grouped keys will be sorted. Default is True.

*Aggregate* with `agg()`

We need to specify an aggregation function, such as `sum()`, `mean()`, `count()`, etc.

Arguments:
- `{'fieldname': 'function', ...}`: pass a dictionary mapping column names to aggregation functions.

Functions include: sum, mean, median, min, max, count, first, last, std, var, etc.

*Flatten column titles* 

The resulting DataFrame from a groupby operation with multiple aggregations will have a MultiIndex for its columns. We can flatten this to a single level by joining the levels with an underscore.

`result.columns = ['_'.join(col) for col in result.columns]`
