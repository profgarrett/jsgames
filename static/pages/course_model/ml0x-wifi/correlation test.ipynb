```python
# Can you tell the correlation?
import random

def generate_correlated_data(n_samples=1000, correlation=0.8, target_p_value=0.01):
    """
    Generate random x and y data with specified correlation and approximate p-value.
    
    Parameters:
    n_samples (int): Number of data points to generate
    correlation (float): Desired correlation coefficient between x and y (-1 to 1)
    target_p_value (float): Target p-value (controls sample size for significance)
    
    Returns:
    pandas.DataFrame: DataFrame with 'x' and 'y' columns
    """
    import numpy as np
    import pandas as pd
    from scipy.stats import pearsonr
    
    # Generate independent random data
    np.random.seed(42)  # For reproducibility
    x = np.random.normal(0, 1, n_samples)
    
    # Generate y with desired correlation
    # y = correlation * x + sqrt(1 - correlation^2) * independent_noise
    independent_noise = np.random.normal(0, 1, n_samples)
    y = correlation * x + np.sqrt(1 - correlation**2) * independent_noise
    
    # Create DataFrame
    df = pd.DataFrame({'x': x, 'y': y})
    
    # Calculate actual correlation and p-value
    actual_corr, actual_p = pearsonr(df['x'], df['y'])
    
    print(f"Generated data:")
    #print(f"  Requested correlation: {correlation:.3f}")
    print(f"  Actual correlation: {actual_corr:.3f}")
    print(f"  Actual p-value: {actual_p:.6f}")
    print(f"  Sample size: {n_samples}")
    
    return df, actual_corr

# Test the function
# Create a random number for the correlation between -1 and 1
correlation = random.uniform(-1, 1)
test_df, actual_corr = generate_correlated_data(n_samples=100, correlation=correlation, target_p_value=0.01)

# Plot the generated data
plt.figure(figsize=(5, 5))
plt.scatter(test_df['x'], test_df['y'], alpha=0.2)
plt.title(f'Sample Correlation {actual_corr:.2f}')
plt.show()

```
